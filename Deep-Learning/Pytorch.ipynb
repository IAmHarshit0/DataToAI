# Neural Networks

```python
import torch.nn
import torch.nn.functional as F
```

```python
class Net(nn.Module):
    
    def __init__(self):
        super(Net, self).__init__()
        self.conv1 = nn.Conv2d(1, 6, 5)
        self.conv2 = nn.Conv2d(6, 16, 5)

        self.fc1 = nn.Linear(16 * 5 * 5, 120)
        self.fc2 = nn.Linear(120, 84)
        self.fc3 = nn.Linear(84,10)

    def forward(self, input):
        c1 = F.relu(self.conv1(input))
        s2 = F.max_pool2d(c1, (2, 2))
        c3 = F.relu(self.conv2(s2))
        s4 = F.max_pool2d(c3, 2)

        s4 = torch.flatten(s4, 1)
        f5 = F.relu(self.fc1(s4))
        f6 = F.relu(self.fc2(f5))

        output = self.fc3(f6)
        return output
    
net = Net()
print(net)
```

## Forward Pass

```python
N = 10
D_in = 1
D_out = 1
```

```python
X = torch.randn(N, D_in)
```

```python
true_W =  torch.tensor([[2.0]])
true_b = torch.tensor(1.0)
```

```python
y_true = X @ true_W + true_b + torch.randn(N, D_out) * 0.1 
```

```python
W = torch.randn(D_in, D_out, requires_grad=True)
b = torch.randn(1, requires_grad=True)
```

```python
print(f"Initial Weight: {W}")
print(f"Initial Bias: {b}")
```

```python
y_hat = X @ W + b 
```

```python
print(f"Shape of our prediction y_hat: {y_hat.shape}\n")
print(f"Prediction y_hat (first 3 rows):\n {y_hat[:3]}\n")
print(f"True Labels y_true (first 3 rows):\n {y_true[:3]}")
```

```python
error = y_hat - y_true
```

```python
squared_error = error ** 2
```

```python
loss = squared_error.mean()
print(loss)
```

```python
loss.backward()
```

```python
W.grad
```

```python
b.grad
```

```python
# Hyperparameters
learning_rate, epochs = 0.01, 100
```

```python
# Reinitailize Parameters
W, b = torch.randn(1, 1, requires_grad=True), torch.randn(1, requires_grad=True)
```

```python
for epoch in range(epochs):
    # Forward Pass And Loss
    y_hat = X @ W + b 
    loss = torch.mean((y_hat- y_true)**2)

    # Backward Loss
    loss.backward()

    # Update Parameters
    with torch.no_grad():
        W -= learning_rate * W.grad
        b -= learning_rate * b.grad

    # Zero Gradients
    W.grad.zero_()
    b.grad.zero_()

    if epoch % 10 == 0:
        print(f'Epoch {epoch:02d}: Loss={loss.item():.4f}, W={W.item():.3f}, b={b.item():.3f}')

print(f"\nFinal Parameters: W={W.item():.3f}, b={b.item():.3f}")
print(f"True Parameters: W=2.000, b=1.000")
```

```python
import torch.nn as nn

class LinearRegressionModel(nn.Module):
    def __init__(self, in_feautres, out_features):
        super().__init__()
        self.linear_layer = nn.Linear(in_features=in_feautres, out_features=out_features)

    def forward(self, x):
        return self.linear_layer(x)
    
model = LinearRegressionModel(in_feautres=1, out_features=1)
```

```python
model
```

```python
import torch.optim as optim

learning_rate = 0.01

optimizer = optim.Adam(model.parameters(), lr=learning_rate)

loss_fn = nn.MSELoss()
```
