```python
import spacy
from spacy.language import Language
from spacy.tokens import Span

nlp = spacy.load("en_core_web_sm")

@Language.component("replace_GPE")
def replace_GPE(doc):
    new_ents = []

    for ent in doc.ents:
        if ent.label_ == "GPE":
            new_ents.append(
                Span(doc, ent.start, ent.end, label="LOC")
            )
        else:
            new_ents.append(ent)

    doc.ents = new_ents
    return doc


# 🔥 THIS IS CRITICAL
nlp.add_pipe("replace_GPE", after="ner")
```

```python
from spacy.language import Language

@Language.component("Apple_MWE_Entity_Ruler")
def apple_mwe_entity_ruler(doc):
    pattern = r"Apple [A-z]\w+"
    orignal_ents = list(doc.ents)
    mwt_ents = []
    for match in re.finditer(pattern, doc.text):
        start, end = match.span()
        span = doc.char_span(start, end)
        if span is not None:
            mwt_ents.append((span.start, span.end, span.text))
            print(span)
    for ent in mwt_ents:
        start, end, name = ent
        per_ent = Span(doc, start, end, label="MULTI_WORD_ENTITY")
        orignal_ents.append(per_ent)
    doc.ents = orignal_ents
    print(doc.ents)
    return (doc)
```

```python
from spacy.language import Language
from spacy.util import filter_spans

@Language.component("Bollywood_NER")
def bollywood_ner(doc):
    pattern = r"Bollywood [A-z]\w+"
    orignal_ents = list(doc.ents)
    mwt_ents = []
    for match in re.finditer(pattern, doc.text):
        start, end = match.span()
        span = doc.char_span(start, end)
        if span is not None:
            mwt_ents.append((span.start, span.end, span.text))
            print(span)
    for ent in mwt_ents:
        start, end, name = ent
        per_ent = Span(doc, start, end, label="BOLLYWOOD_ENTITY")
        orignal_ents.append(per_ent)
    filtered = filter_spans(orignal_ents)
    doc.ents = filtered
    print(doc.ents)
    return (doc)
```

```python
nlp3 = spacy.load('en_core_web_sm')
nlp3.add_pipe("Bollywood_NER")
```

```python
text = "Bollywood Superstar Shah Rukh Khan is known as the King of Bollywood"
doc3 = nlp3(text)
for ent in doc3.ents:
    print(ent.text, ent.label_)
```
