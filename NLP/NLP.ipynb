# Bag Of Words

```python
from sklearn.feature_extraction.text import CountVectorizer

train_data = [
    "I love programming in Python",
    "i love coding in Java",
    "Python and Java are popular programming languages"
]

vectorizer = CountVectorizer()
vectors = vectorizer.fit_transform(train_data)

print(vectorizer.get_feature_names_out())
print(vectors.toarray())
```

```python
from sklearn.feature_extraction.text import CountVectorizer

train_data = [
    "I love programming in Python python",
    "i love coding in Java",
    "Python and Java are popular programming languages",
    "HOML is a great programming book",
]

vectorizer = CountVectorizer()
vectors = vectorizer.fit_transform(train_data)

print(vectorizer.get_feature_names_out())
print(vectors.toarray())
```

```python
class category:
    PROGRAMMING = "PROGRAMMING"
    BOOK = "BOOK"

train_y = [
    category.PROGRAMMING,
    category.PROGRAMMING,
    category.BOOK,
    category.BOOK]
```

```python
from sklearn import svm

clf = svm.SVC()
clf.fit(vectors, train_y)
```

```python
test_x = vectorizer.transform([
    "I enjoy writing code in Python"])

clf.predict(test_x)
```

```python
test_x = vectorizer.transform(["HOML is a great programming book"])

clf.predict(test_x)
```

## Bigrams

```python
from sklearn.feature_extraction.text import CountVectorizer

train_data = [
    "I love programming in Python python",
    "i love coding in Java",
    "Python and Java have some great books",
    "HOML is a great programming book",
]

vectorizer = CountVectorizer(binary=True, ngram_range=(1,2))
vectors = vectorizer.fit_transform(train_data)

print(vectorizer.get_feature_names_out())
print(vectors.toarray())
```

# Regex

```python
import re 

regexp = re.compile(r"^ab[^\s]*cd$")
phrases = ["abxcd", "xxx", "zz xabcdx zz", "ab cd"]

matches = []
for phrase in phrases:
    if re.match(regexp, phrase):
        matches.append(phrase)
print(matches)  
```

```python
import re 

regexp = re.compile(r"ab[^\s]*cd")
phrases = ["abxcd", "xxx", "zz abxxcd zz", "ab cd"]

matches = []
for phrase in phrases:
    if re.search(regexp, phrase):
        matches.append(phrase)
print(matches)  
```

```python
import re

regexp = re.compile(r"bread|story|book")
phrases = ["bread", "story", "book", "bread book"]

matches = []
for phrase in phrases:
    if re.search(regexp, phrase):
        matches.append(phrase)
print(matches)
```

```python
import re

regexp = re.compile(r"\bbread\b|\bstory\b|\bbook\b")
phrases = ["breaded", "history", "booked", "bread book"]

matches = []
for phrase in phrases:
    if re.search(regexp, phrase):
        matches.append(phrase)
print(matches)
```

# Stemming/Lemmatization

```python
import nltk

nltk.download('wordnet')
nltk.download('stopwords')
nltk.download('punkt')
```

```python
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

phrase = "reading readers readily read readable readings"
words = word_tokenize(phrase)

stemmed_words = []
for word in words:
    stemmed_words.append(stemmer.stem(word))
print(" ".join(stemmed_words))
```

```python
from nltk.stem import WordNetLemmatizer
 
lemmatizer = WordNetLemmatizer()

lemma_words = [lemmatizer.lemmatize(word, pos="v") for word in words]

print(" ".join(lemma_words))
```

# Stopwords Removal

```python
from nltk.corpus import stopwords

stopwords = stopwords.words('english')

phrase = "This is a sample sentence, showing off the stop words filtration."

words = word_tokenize(phrase)

stripped_phrase = []
for word in words:
    if word.lower() not in stopwords:
        stripped_phrase.append(word)
print(" ".join(stripped_phrase))
```
