## Hypothesis Testing

### Z Test

```python
# Example 1
s_x = 90
p_u = 82
p_std = 20
n = 81
ap = 0.05
```

```python
z_cal = (s_x - p_u)/(p_std/np.sqrt(n))
z_cal
```

```python
z_table = st.norm.ppf(1 - ap)
z_table
```

```python
if z_table < z_cal:
    print("Ha is right")
else:
    print("H0 is right")
```

### T Test

```python
# Example 1
pop_u = 150 
sam_x = 148
n = 25
sam_std = 5
ap = 0.05
df = n-1
```

```python
t_cal = (sam_x - pop_u)/(sam_std/np.sqrt(n))
t_cal
```

```python
t_table = st.t.ppf(1-ap, df)
t_table
```

```python
if (np.abs(t_cal) > t_table):
    print("Ha is right")
else:
    print("H0 is right")
```

### Chi Squared Test

```python
# Example 1
ob = np.array([22, 17, 20, 26, 22, 13])
ex = np.array([20,20,20,20,20,20])
```

```python
chi_obs = np.sum(np.square(ob-ex)/ex)
chi_obs 
```

```python
# Example 2
row1 = np.array([40,45,25,10])
row2 = np.array([35,30,20,30])
```

```python
sum_r1 = np.sum(row1)
sum_r2 = np.sum(row2)
sum_row = np.array([sum_r1, sum_r2])
```

```python
sum_cal = row1 + row2
sum_cal
```

```python
exp = []
for i in sum_row:
    for j in sum_cal:
        exp.append((i*j)/235)
exp
```

```python
obj = np.array([40,45,25,10,35,30,20,30])
obj
```

```python
chi_cal = np.sum(np.square(obj-exp)/exp)
chi_cal
```

```python
chi_table = 7.815
```

```python
if chi_cal > chi_table:
    print("Ha is right")
else:
    print("H0 is right")
```
