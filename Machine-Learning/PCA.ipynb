```python
import matplotlib.pyplot as plt
import pandas as pd 
import numpy as np 
import seaborn as sns 
```

```python
from sklearn.datasets import load_breast_cancer
```

```python
cancer = load_breast_cancer()
```

```python
cancer.keys()
```

```python
print(cancer['DESCR'])
```

```python
df = pd.DataFrame(cancer['data'], columns=cancer['feature_names'])
```

```python
df.head()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
ss = StandardScaler()
```

```python
ss.fit(df)
```

```python
scaler_data = ss.transform(df)
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2)
```

```python
pca.fit(scaler_data)
```

```python
x_pca = pca.transform(scaler_data)
```

```python
scaler_data.shape
```

```python
x_pca.shape
```

```python
plt.scatter(x_pca[:,0], x_pca[:,1], c=cancer['target'],cmap='plasma')
plt.xlabel("First Component Analysis")
plt.ylabel("Second Component Analysis")
plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
mm = MinMaxScaler()
```

```python
mm.fit(df)
```

```python
mm_dataset = mm.transform(df)
```

```python
pca.fit(mm_dataset)
```

```python
x_pca_mm = pca.transform(mm_dataset)
```

```python
mm_dataset.shape
```

```python
x_pca_mm.shape
```

```python
plt.scatter(x_pca_mm[:,0], x_pca_mm[:,1], c=cancer['target'],cmap='plasma')
plt.xlabel("First Component Analysis")
plt.ylabel("Second Component Analysis")
plt.show()
```
